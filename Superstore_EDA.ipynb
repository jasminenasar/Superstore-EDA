```python
import matplotlib.pyplot as plt

profit_by_subcat = df.groupby('Sub-Category')['Profit'].sum().sort_values()

plt.figure(figsize=(10,6))
colors = ['red' if x < 0 else 'green' for x in profit_by_subcat]
plt.barh(profit_by_subcat.index, profit_by_subcat.values, color=colors)
plt.xlabel('Total Profit ($)')
plt.title('Profit by Sub-Category')
plt.axvline(0, color='black', linewidth=0.8)
plt.tight_layout()
plt.show()
```

```python
df['Order Month'] = df['Order Date'].dt.to_period('M')
monthly = df.groupby('Order Month')[['Sales','Profit']].sum()
monthly
```

```python
monthly.index = monthly.index.to_timestamp()

plt.figure(figsize=(12,6))
plt.plot(monthly.index, monthly['Sales'], label='Sales', marker='o')
plt.plot(monthly.index, monthly['Profit'], label='Profit', marker='o')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.title('Monthly Sales & Profit Trend')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Superstore Sales & Profit — EDA Summary

Dataset: 10,194 orders, 21 columns, spanning Jan 2023 – Dec 2026, no missing values.
    
Key Findings:
    
1.Discounting hurts profit. Discount and Profit have a negative correlation (-0.22). Orders with discounts above ~20-25% frequently turn unprofitable, and by 50% discount, average profit is -$310 per order. One order — a discounted 3D printer at 70% off — lost $6,599 on its own.

2.Furniture underperforms despite strong sales. Furniture generated $754K in sales but only $19.7K profit — far below Office Supplies ($126K profit) and Technology ($146.5K profit) on similar sales volume.

3.Tables and Bookcases are losing money. Tables lost -$17,753 overall despite $208K in sales — the single worst-performing sub-category. Bookcases (-$3,632) and Supplies (-$1,171) also ran at a loss. The likely cause: Tables carry a 25.8% average discount, versus just 15.7% for Copiers (the most profitable sub-category).

4.Copiers, Phones, and Accessories are the most profitable sub-categories, led by Copiers at $56K profit on $150K in sales — the best margin in the dataset.

5.Sales grew significantly over time (from ~$10-30K/month in 2023 to peaks over $100K/month by 2026), but profit stayed comparatively flat throughout — revenue growth did not translate into proportional profit growth, consistent with the discounting pattern above.
```
